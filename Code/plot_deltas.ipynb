```python
import bagpipes as pipes
import numpy as np

burst1 = {}
burst1["age"] = 0.50          # Gyr before observation
burst1["massformed"] = 9.0    # log10(M*/Msun)
burst1["metallicity"] = 0.5

burst2 = {}
burst2["age"] = 1.5        # 0.50 + 0.06 Gyr
burst2["massformed"] = 9.0
burst2["metallicity"] = 0.5

dust = {}
dust["type"] = "Calzetti"
dust["Av"] = 0.2

model_components = {}
model_components["redshift"] = 2.0
model_components["burst1"] = burst1
model_components["burst2"] = burst2
model_components["dust"] = dust
```

```python
base = (
    r"D:/Users/Abigail Crowther/Desktop/Uni/Year 4/MPhys Projects/"
    r"Semester 2 - Star Formation Histories/star-formation-histories/bagpipes/examples"
)

rel_filters = np.loadtxt(base + "/filters/goodss_filt_list.txt", dtype="str")
goodss_filt_list = [base + "/" + f for f in rel_filters]

model = pipes.model_galaxy(model_components, filt_list=goodss_filt_list)
```

```python
fig = model.plot()
fig = model.sfh.plot()
```
