```python
import numpy as np
import bagpipes as pipes

def load_dummy(ID):
    return np.zeros((len(filter_files), 2))

dummy_fit_info = {
    "sfh": "exponential",
    "metallicity": 0.02,
    "dust": {"type": "Calzetti", "Av": 0.0},
    "redshift": z,      
    "massformed": 1e10
}

fits = []

for t in years:
    galaxy = pipes.galaxy(
        f"{t:.1f}",
        load_dummy,
        spectrum_exists=False,
        photometry_exists=True,
        filt_list=filter_files
    )

    fit = pipes.fit(
        galaxy,
        dummy_fit_info,
        run=f"exponential/z_{z}",               
    )

    fits.append(fit)

print(f"Loaded {len(fits)} existing fit objects")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from bagpipes import utils

z_model = z
M_target = 1e10   # desired total formed mass

n_rows, n_cols = 5, 4
fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 15), sharex=True, sharey=True)
axes = axes.flatten()

for i, burst_age in enumerate(years):

    ax = axes[i]

    # cosmic age at model redshift 
    age_universe = np.interp(z_model, utils.z_array, utils.age_at_z)

    n_bins = 1000
    ages_array = np.linspace(0, age_universe, n_bins)

    # Gaussian burst shape (unnormalised)
    sigma = (ages_array.max() - ages_array.min()) * 0.01
    burst_center = age_universe - burst_age
    sfr_unnorm = np.exp(-(ages_array - burst_center)**2 / (2 * sigma**2))

    # normalise to total mass M_target
    integral = np.trapz(sfr_unnorm, ages_array * 1e9)  # convert Gyr→yr
    scale = M_target / integral
    sfr_array = scale * sfr_unnorm

    ax.plot(ages_array, sfr_array, color="blue",
            label=f"True Burst {burst_age:.2f} Gyr")

    fit = fits[i]

    if "redshift" in fit.fitted_model.params:
        z_fit = np.median(fit.posterior.samples["redshift"])
    else:
        z_fit = fit.fitted_model.model_components["redshift"]

    age_universe_fit = np.interp(z_fit, utils.z_array, utils.age_at_z)

    post = np.percentile(fit.posterior.samples["sfh"], (16, 50, 84), axis=0).T

    # Bagpipes ages = lookback time in years → convert to cosmic age (Gyr)
    lookback = fit.posterior.sfh.ages * 1e-9
    cosmic_age = age_universe_fit - lookback

    ax.plot(cosmic_age, post[:, 1], color="red", alpha=0.7, label="Recovered SFH")
    ax.fill_between(cosmic_age, post[:, 0], post[:, 2], color="red", alpha=0.2)

    ax.set_xlim(age_universe_fit, 0)
    ax.set_ylim(0, 370)
    ax.set_xlabel("Age of Universe (Gyr)")
    ax.set_ylabel(r"SFR ($M_\odot\,\mathrm{yr}^{-1}$)")
    ax.legend(fontsize=8)

    if i < n_cols:
        ax_top = ax.twiny()
        zvals = [0, 0.5, 1, 2, 4, 10]
        ages_top = np.interp(zvals, utils.z_array, utils.age_at_z)
        ax_top.set_xlim(ax.get_xlim())
        ax_top.set_xticks(ages_top)
        ax_top.set_xticklabels(zvals)
        ax_top.set_xlabel("Redshift")

# remove unused axes
for j in range(i+1, n_rows*n_cols):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
